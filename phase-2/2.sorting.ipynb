# sorting numpy array
### In NumPy, sorting an array means arranging its elements in ascending or descending order.

## 1)sorting one D array 

## a)in ascending order




### np.sort()

```python
import numpy as np

org_arr=np.array([20,40,60,10])

new_sorted_arr=np.sort(org_arr) # it will return new sorted array ,doesn't sort original array

print(f"sorting one D array:{new_sorted_arr}") 
print(f"original array:{org_arr}")
```

### sorting original array

```python
import numpy as np
np_arr=np.array([10,20,30,40])
np_arr.sort()  # it will sort original array
print(np_arr)
```

## conclustion 

### np.sort(arr)  → returns a new sorted array
### arr.sort()    → modifies the original array

### by default it sort in ascending order

# b) in descending order 

### np.sort()

```python
import numpy as np 
num_arr=np.array([10,20,30,40])
new_sorted_array=np.sort(num_arr)
descending_array=new_sorted_array[::-1]
print(descending_array)
```

## sorting original array 

```python
import numpy as np 
num_arr=np.array([10,20,30,40])
num_arr.sort()
print(num_arr[::-1])
```
